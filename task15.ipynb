```
double myStep = 1e-4; 
int[] threadCounts = { 1, 2, 4, 6, 8, 12, 16 };
double[] results = new double[threadCounts.Length];
Stopwatch sw = new();

sw.Start();
DefiniteIntegral.SolveSingleThread(-100, 100, Math.Sin, myStep);
sw.Stop();
double singleTime = sw.Elapsed.TotalMilliseconds;

Console.WriteLine($"~~~ РЕЗУЛЬТАТЫ ЗАМЕРОВ ~~~");
Console.WriteLine($"Однопоточная версия (без потоков): {singleTime:F2} мс");

for(int i = 0; i < threadCounts.Length; i++) {
    List<double> runs = new();
    for(int j = 0; j < 5; j++) { 
        sw.Restart();
        DefiniteIntegral.Solve(-100, 100, Math.Sin, myStep, threadCounts[i]);
        sw.Stop();
        runs.Add(sw.Elapsed.TotalMilliseconds);
    }
    results[i] = runs.Average();
    Console.WriteLine($"Потоков: {threadCounts[i]}  Среднее время: {results[i]:F2} мс");
}

var plt = new ScottPlot.Plot();
var scatter = plt.Add.Scatter(results, threadCounts.Select(x => (double)x).ToArray());
scatter.LegendText = "Многопоточность";

var baseline = plt.Add.VerticalLine(singleTime); 
baseline.Color = ScottPlot.Colors.Red;
baseline.LegendText = "Один поток";     
baseline.LineStyle.Pattern = ScottPlot.LinePattern.Dashed;

plt.Title("Эффективность многопоточности");
plt.XLabel("Время вычисления"); 
plt.YLabel("Количество потоков");  

plt.ShowLegend();
plt.Display();
plt.SavePng("result.png", 800, 600);

```
